# Session 05 — Color Filtering and Masks

Run locally with a webcam. Tune an HSV mask, clean it with morphology, and follow the largest connected component.


```
%pip install opencv-python
import cv2
import numpy as np
```

```
# EDIT THESE HSV values after inspecting the trackbars.
LOWER=np.array([0,120,80])
UPPER=np.array([20,255,255])
KERNEL=np.ones((5,5),np.uint8)

camera=cv2.VideoCapture(0)
if not camera.isOpened(): raise RuntimeError("No webcam available. Run locally and allow camera access.")
while True:
    ok,frame=camera.read()
    if not ok: break
    hsv=cv2.cvtColor(frame,cv2.COLOR_BGR2HSV)
    mask=cv2.inRange(hsv,LOWER,UPPER)
    cleaned=cv2.morphologyEx(mask,cv2.MORPH_OPEN,KERNEL)
    cleaned=cv2.morphologyEx(cleaned,cv2.MORPH_CLOSE,KERNEL)
    count,labels,stats,centroids=cv2.connectedComponentsWithStats(cleaned)
    if count>1:
        i=1+np.argmax(stats[1:,cv2.CC_STAT_AREA])
        x,y,w,h,area=stats[i]
        cx,cy=centroids[i]
        cv2.rectangle(frame,(x,y),(x+w,y+h),(21,138,194),2)
        cv2.circle(frame,(int(cx),int(cy)),5,(246,138,69),-1)
        cv2.putText(frame,f'area: {area}',(x,max(24,y-8)),cv2.FONT_HERSHEY_SIMPLEX,.6,(255,255,255),2)
    cv2.imshow('Camera',frame);cv2.imshow('HSV mask',cleaned)
    if cv2.waitKey(1)&0xFF==ord('q'):break
camera.release();cv2.destroyAllWindows()
```
