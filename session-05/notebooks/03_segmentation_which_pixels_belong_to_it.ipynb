# Session 05 — Segmentation

## Segmentation — Which pixels belong to each object?


**Mission**

Segmentation marks the visible pixels belonging to an object. A detection box only marks an area around it.


```python
# Install Ultralytics because it supplies the pretrained segmentation model.
!pip -q install ultralytics
```

```python
# urlretrieve downloads the public sample image.
from urllib.request import urlretrieve
# display renders images below notebook cells.
from IPython.display import display
# Image opens input images and wraps model output for display.
from PIL import Image
# YOLO loads the pretrained segmentation model.
from ultralytics import YOLO
```

### Predict

Pick one object in the image. Where should its boundary stop?


```python
# IMAGE_URL is the source image; replace it to test another scene.
IMAGE_URL = "https://ultralytics.com/images/bus.jpg"
# IMAGE_PATH is the downloaded local filename.
IMAGE_PATH = "scene.jpg"
# Save the remote image locally so later cells use the same input.
urlretrieve(IMAGE_URL, IMAGE_PATH)
```

```python
# Show the original RGB image before looking at its predicted masks.
display(Image.open(IMAGE_PATH).convert("RGB"))
```

### Run segmentation

Start with a confidence level, then compare the masks with the object boundaries you predicted.


```python
# CONFIDENCE is the minimum score required for a predicted object mask.
CONFIDENCE = 0.35
# Load YOLO11's lightweight segmentation weights.
model = YOLO("yolo11n-seg.pt")
```

```python
# Run segmentation and select the result for the first input image.
result = model(IMAGE_PATH, conf=CONFIDENCE, verbose=False)[0]
# result.plot() overlays masks; reversing channels converts BGR output to RGB for display.
annotated = Image.fromarray(result.plot()[:, :, ::-1])
# Show the object masks generated by the model.
display(annotated)
```

```python
# result.masks is None when no object receives a segmentation mask.
# len(result.masks.data) counts the masks when masks are present.
mask_count = 0 if result.masks is None else len(result.masks.data)
print(f"Objects with masks: {mask_count}")
```
