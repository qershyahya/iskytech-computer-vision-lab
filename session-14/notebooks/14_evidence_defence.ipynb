# Session 14 — Evidence Defence

**Goal:** A result is defendable when baseline, change, output, and limitation are explicit.

Run the baseline first. Change only the marked value or line, then record what the output supports and what it does not.


## Setup


```python
from pathlib import Path
from urllib.request import urlretrieve
from PIL import Image, ImageEnhance, ImageFilter
from IPython.display import display
```

```python
%pip -q install ultralytics
from ultralytics import YOLO
IMAGE_URL='http://images.cocodataset.org/val2017/000000039769.jpg'
path=Path('cats.jpg')
if not path.exists(): urlretrieve(IMAGE_URL,path)
original=Image.open(path).convert('RGB')
display(original)
```

## Baseline


```python
model=YOLO('yolo11n.pt')
baseline=model(original,conf=0.25,verbose=False)[0]
print(f'Baseline detections: {len(baseline.boxes)}')
display(Image.fromarray(baseline.plot()[:,:,::-1]))
```

## Evidence defence


```python
# Compare the baseline with one stress condition.
BLUR_RADIUS=8
stress=original.filter(ImageFilter.GaussianBlur(BLUR_RADIUS))
stress_result=model(stress,conf=0.25,verbose=False)[0]
print({'baseline_detections':len(baseline.boxes),'stress_detections':len(stress_result.boxes),'blur_radius':BLUR_RADIUS})
display(Image.fromarray(stress_result.plot()[:,:,::-1]))
```

## Evidence record

Record the input, exact change or rule, observed output, failure point or limitation, and the next test.
