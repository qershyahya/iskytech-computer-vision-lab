## 2. Run the working baseline

This new challenge image contains cats. Run this cell before editing anything. It gives you the result that you will try to break.


```python
# Keep this source fixed so that every learner starts from the same evidence.
IMAGE_URL = 'http://images.cocodataset.org/val2017/000000039769.jpg'
image_path = Path('challenge-cats.jpg')
if not image_path.exists():
    urlretrieve(IMAGE_URL, image_path)

original = Image.open(image_path).convert('RGB')
print(f'Baseline image: {original.size}')
display(original)

# Run the working baseline first. Save its labels and confidence in your notes.
model = YOLO('yolo11n.pt')
baseline = model(original, conf=0.25, verbose=False)[0]
baseline_image = Image.fromarray(baseline.plot()[:, :, ::-1])
print(f'Baseline detections: {len(baseline.boxes)}')
display(baseline_image)
```
