## Classification — What is in this image?

**Mission**

A classifier gives labels for the whole image. It does not tell us where an object is.

```python
# urlretrieve downloads a public sample image without a separate package.
from urllib.request import urlretrieve
# display renders an image directly below a notebook cell.
from IPython.display import display
# Image opens and converts the downloaded image into RGB pixels.
from PIL import Image
# YOLO loads and runs the pretrained computer-vision model.
from ultralytics import YOLO
```

### Predict

Before running code: what do you think the model will call the main subject of this image?

```python
# IMAGE_URL is the input image address. Replace it to run the same experiment on another image.
IMAGE_URL = "https://ultralytics.com/images/bus.jpg"
# IMAGE_PATH is the local filename created after the download.
IMAGE_PATH = "scene.jpg"
# urlretrieve(source, destination) saves the remote image to the local filename.
urlretrieve(IMAGE_URL, IMAGE_PATH)
```

```python
# Open the local file and convert it to RGB so red, green, and blue use the expected order.
image = Image.open(IMAGE_PATH).convert("RGB")
# Show the exact input that the model will classify.
display(image)
```

### Inspect the result

Will the top label describe the whole scene, one object, or something else?

```python
# Load YOLO11's lightweight image-classification weights.
model = YOLO("yolo11n-cls.pt")
# Run the model once; result.probs stores one probability for each possible label.
result = model(IMAGE_PATH, verbose=False)[0]
```

```python
# TOP_K controls how many of the highest-probability labels are shown.
TOP_K = 5
# enumerate adds a human-friendly ranking number, beginning at 1.
for rank, class_id in enumerate(result.probs.top5[:TOP_K], start=1):
    # result.names converts the numeric class ID into a readable label.
    label = result.names[class_id]
    # result.probs.data[class_id] is the model probability for this label.
    confidence = float(result.probs.data[class_id])
    # :.1% changes a decimal probability, such as 0.842, into 84.2%.
    print(f"{rank}. {label}: {confidence:.1%}")
```
