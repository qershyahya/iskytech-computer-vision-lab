# Session 08 — From Classification to Detection

**Goal:** run a pretrained detector and interpret classes, boxes, confidence, and non-maximum suppression.


```
%pip -q install ultralytics
from pathlib import Path
from urllib.request import urlretrieve
from PIL import Image
from IPython.display import display
from ultralytics import YOLO
```

```
URL='http://images.cocodataset.org/val2017/000000039769.jpg'
path=Path('cats.jpg')
if not path.exists():urlretrieve(URL,path)
image=Image.open(path).convert('RGB');display(image)
model=YOLO('yolo11n.pt')
```

```
# EDIT: confidence removes weak boxes; IOU controls NMS overlap suppression.
CONFIDENCE=.25
NMS_IOU=.70
result=model(image,conf=CONFIDENCE,iou=NMS_IOU,verbose=False)[0]
print(f'Boxes after NMS: {len(result.boxes)} | confidence={CONFIDENCE} | NMS IoU={NMS_IOU}')
display(Image.fromarray(result.plot()[:,:,::-1]))
for box in result.boxes:
 print(result.names[int(box.cls[0])],f'{float(box.conf[0]):.1%}',box.xyxy[0].tolist())
```
