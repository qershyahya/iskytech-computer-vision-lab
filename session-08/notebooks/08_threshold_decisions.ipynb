# Session 08 — Threshold Decisions

**Goal:** A threshold trades weak extra boxes against missing real objects.

Run the baseline first. Change only the marked value or line, then record what the output supports and what it does not.


## Setup


```python
from pathlib import Path
from urllib.request import urlretrieve
from PIL import Image, ImageEnhance, ImageFilter
from IPython.display import display
```

```python
%pip -q install ultralytics
from ultralytics import YOLO
IMAGE_URL='http://images.cocodataset.org/val2017/000000039769.jpg'
path=Path('cats.jpg')
if not path.exists(): urlretrieve(IMAGE_URL,path)
original=Image.open(path).convert('RGB')
display(original)
```

## Baseline


```python
model=YOLO('yolo11n.pt')
baseline=model(original,conf=0.25,verbose=False)[0]
print(f'Baseline detections: {len(baseline.boxes)}')
display(Image.fromarray(baseline.plot()[:,:,::-1]))
```

## Threshold test


```python
# EDIT THIS VALUE: compare 0.10, 0.60, and 0.85.
CONFIDENCE=0.60
result=model(original,conf=CONFIDENCE,verbose=False)[0]
print(f'Threshold {CONFIDENCE}: {len(result.boxes)} retained boxes')
display(Image.fromarray(result.plot()[:,:,::-1]))
```
