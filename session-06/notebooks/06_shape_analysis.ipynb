# Session 06 — Object Detection and Shape Analysis

**Goal:** use contours and geometry to count, classify, and measure simple objects, then compare the method with deep-learning detection.


```
%pip install opencv-python matplotlib
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```
# A controlled test image with one circle, rectangle, and triangle.
image=np.full((420,640,3),255,dtype=np.uint8)
cv2.circle(image,(120,210),65,(0,0,0),-1)
cv2.rectangle(image,(260,145),(400,275),(0,0,0),-1)
cv2.fillPoly(image,[np.array([[500,285],[565,145],[620,285]])],(0,0,0))
# EDIT THIS: changing the threshold tests the boundary rule.
THRESHOLD=127
gray=cv2.cvtColor(image,cv2.COLOR_BGR2GRAY)
_,binary=cv2.threshold(gray,THRESHOLD,255,cv2.THRESH_BINARY_INV)
contours,_=cv2.findContours(binary,cv2.RETR_EXTERNAL,cv2.CHAIN_APPROX_SIMPLE)
result=image.copy()
for contour in contours:
    area=cv2.contourArea(contour)
    perimeter=cv2.arcLength(contour,True)
    approx=cv2.approxPolyDP(contour,.04*perimeter,True)
    corners=len(approx)
    name='triangle' if corners==3 else 'rectangle' if corners==4 else 'circle'
    x,y,w,h=cv2.boundingRect(contour)
    cv2.rectangle(result,(x,y),(x+w,y+h),(21,138,194),2)
    cv2.putText(result,f'{name}: {area:.0f}px', (x,y-10),cv2.FONT_HERSHEY_SIMPLEX,.5,(246,138,69),2)
print('Objects counted:',len(contours))
plt.figure(figsize=(10,5));plt.imshow(cv2.cvtColor(result,cv2.COLOR_BGR2RGB));plt.axis('off');plt.show()
```
