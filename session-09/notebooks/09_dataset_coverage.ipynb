# Session 09 — Dataset Coverage

**Goal:** A model cannot prove it works in conditions nobody tested.

Run the baseline first. Change only the marked value or line, then record what the output supports and what it does not.


## Setup


```python
from pathlib import Path
from urllib.request import urlretrieve
from PIL import Image, ImageEnhance, ImageFilter
from IPython.display import display
```

## Inventory a test set


```python
# Each entry names an image condition you have actually tested.
test_set=[{'file':'bright_cat.jpg','light':'bright','blur':'none','crop':'none'}, {'file':'dark_cat.jpg','light':'dark','blur':'none','crop':'none'}]
for item in test_set: print(item)
```

## Find the gap


```python
# EDIT THIS LIST after inspecting your own test set.
required_conditions={'light':{'bright','dark'},'blur':{'none','high'},'crop':{'none','partial'}}
for key,expected in required_conditions.items():
    observed={item[key] for item in test_set}
    print(key,'missing:',sorted(expected-observed))
```
