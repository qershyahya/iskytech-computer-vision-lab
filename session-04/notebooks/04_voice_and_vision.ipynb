# Session 04 — Voice and Vision

**Goal:** turn a voice or text command into a validated image-processing action and visible feedback. Text input is the dependable stand-in for a speech recogniser; a speech recogniser can pass its transcript into the same function.


```
from pathlib import Path
from urllib.request import urlretrieve
from PIL import Image, ImageOps, ImageFilter
from IPython.display import display

IMAGE_URL='http://images.cocodataset.org/val2017/000000039769.jpg'
path=Path('cats.jpg')
if not path.exists(): urlretrieve(IMAGE_URL,path)
original=Image.open(path).convert('RGB')
display(original)
```

## Validate before acting


```
# Commands from voice or text are normalised into this small allow-list.
ACTIONS={'original':'original','gray':'gray','grayscale':'gray','edges':'edges','threshold':'threshold'}

def run_command(command, image):
    key=command.strip().lower()
    mode=ACTIONS.get(key)
    if mode is None:
        return image, f"Rejected: {command!r}. Try: {', '.join(sorted(ACTIONS))}."
    if mode=='original': return image, 'Mode: original'
    if mode=='gray': return ImageOps.grayscale(image).convert('RGB'), 'Mode: grayscale'
    if mode=='edges': return image.filter(ImageFilter.FIND_EDGES), 'Mode: edges'
    if mode=='threshold':
        gray=ImageOps.grayscale(image)
        return gray.point(lambda pixel: 255 if pixel>125 else 0).convert('RGB'), 'Mode: threshold'

# EDIT THIS: substitute a recognised speech transcript here.
COMMAND='edges'
result,status=run_command(COMMAND,original)
print(status)
display(result)
```
