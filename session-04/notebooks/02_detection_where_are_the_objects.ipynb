## Object Detection — Where are the objects?


**Mission**

A detector returns an object label, a confidence score, and a bounding box for each finding.


```python
# Install Ultralytics because it supplies the pretrained object detector.
!pip -q install ultralytics
```

```python
# urlretrieve downloads the public sample image.
from urllib.request import urlretrieve
# display places images directly below a cell.
from IPython.display import display
# Image opens images and converts model output into a notebook-displayable image.
from PIL import Image
# YOLO loads and runs the pretrained detector.
from ultralytics import YOLO
```

### Predict

List the objects you expect the model to find. Which one might it miss?


```python
# IMAGE_URL is the source image. Change it to test another scene.
IMAGE_URL = "http://images.cocodataset.org/val2017/000000039769.jpg"
# IMAGE_PATH is the local name used by the later model call.
IMAGE_PATH = "cats.jpg"
# Download the source image before displaying or analysing it.
urlretrieve(IMAGE_URL, IMAGE_PATH)
```

```python
# Convert the input to RGB and show the scene before asking the detector for answers.
display(Image.open(IMAGE_PATH).convert("RGB"))
```

### Set a decision rule

Choose a starting confidence. A higher value demands stronger evidence before a box appears.


```python
# CONFIDENCE is the minimum score required before the detector draws a box.
CONFIDENCE = 0.35
# Load YOLO11's lightweight object-detection weights.
model = YOLO("yolo11n.pt")
```

```python
# Run detection with the selected minimum confidence and take the first image result.
result = model(IMAGE_PATH, conf=CONFIDENCE, verbose=False)[0]
# result.plot() draws the model's boxes in BGR order; [ :, :, ::-1 ] reverses channels to RGB.
annotated = Image.fromarray(result.plot()[:, :, ::-1])
# Display the boxes, labels, and confidence values drawn by the model.
display(annotated)
```

### Inspect the evidence

Which box has the weakest confidence? Does it look convincing to you?


```python
# result.boxes contains one bounding-box record for every retained detection.
for box in result.boxes:
    # box.cls[0] is this box's numeric category ID.
    class_id = int(box.cls[0])
    # result.names translates the category ID into a readable object name.
    label = result.names[class_id]
    # box.conf[0] is this individual box's confidence score.
    confidence = float(box.conf[0])
    print(f"{label} — confidence {confidence:.1%}")
```
