# Session 07 — Deep Learning in Vision

Run a pretrained classifier, inspect confidence, then apply a realistic augmentation.


```
%pip -q install ultralytics
from pathlib import Path
from urllib.request import urlretrieve
from PIL import Image, ImageEnhance
from IPython.display import display
from ultralytics import YOLO
```

```
URL='http://images.cocodataset.org/val2017/000000039769.jpg'
path=Path('cats.jpg')
if not path.exists():urlretrieve(URL,path)
original=Image.open(path).convert('RGB');display(original)
model=YOLO('yolo11n-cls.pt')
result=model(original,verbose=False)[0]
for rank,class_id in enumerate(result.probs.top5[:3],1):print(rank,result.names[class_id],f'{float(result.probs.data[class_id]):.1%}')
```

```
# EDIT THIS: realistic light augmentation; compare to baseline.
LIGHT_FACTOR=.45
augmented=ImageEnhance.Brightness(original).enhance(LIGHT_FACTOR)
aug_result=model(augmented,verbose=False)[0]
print('Augmented top label:',aug_result.names[aug_result.probs.top1],f'{float(aug_result.probs.top1conf):.1%}')
display(augmented)
```
