# Session 10 — Robustness Testing

**Goal:** A useful test changes one condition and compares it to a recorded baseline.

Run the baseline first. Change only the marked value or line, then record what the output supports and what it does not.


## Setup


```python
from pathlib import Path
from urllib.request import urlretrieve
from PIL import Image, ImageEnhance, ImageFilter
from IPython.display import display
```

```python
%pip -q install ultralytics
from ultralytics import YOLO
IMAGE_URL='http://images.cocodataset.org/val2017/000000039769.jpg'
path=Path('cats.jpg')
if not path.exists(): urlretrieve(IMAGE_URL,path)
original=Image.open(path).convert('RGB')
display(original)
```

## Baseline


```python
model=YOLO('yolo11n.pt')
baseline=model(original,conf=0.25,verbose=False)[0]
print(f'Baseline detections: {len(baseline.boxes)}')
display(Image.fromarray(baseline.plot()[:,:,::-1]))
```

## One controlled stress test


```python
# EDIT ONE VALUE. Keep the baseline unchanged.
BLUR_RADIUS=8
working=original.filter(ImageFilter.GaussianBlur(BLUR_RADIUS))
result=model(working,conf=0.25,verbose=False)[0]
print(f'Blur {BLUR_RADIUS}: {len(result.boxes)} detections')
display(working)
display(Image.fromarray(result.plot()[:,:,::-1]))
```
